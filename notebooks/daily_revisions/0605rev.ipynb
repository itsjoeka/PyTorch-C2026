**MY INTUITION FOR CREATING N-DIMENSIONAL ARRAYS AND TENSORS**

- The `square bracket` theory
  - **Scalars** are created with no square bracket     : `torch.tensor()`
    - 0 dimensions
  - **Vectors** are created with one square bracket    : `torch.tensor([])`
    - square bracket contains `scalars`
    - 1 dimension: takes only rows, or only columns
  - **MATRICES** are created with two square brcakets  : `torch.tensor([  []  ])`
    - square brackets contains `vectors`
    - 2 dimensions: takes both rows and columns
  - **TENSORS** are created with three square brackers : `torch.tensor([ [[]] ])`
    - square bracket contains `matrices`
    - 3 dimensions: takes multiple layers for each matrix
