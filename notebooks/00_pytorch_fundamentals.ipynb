### Random tensors

Why random tensors?

Random tensors are important because the way many neural networks learn is that they start with tensors full of random numbers and then adjust those random numbers to better represent the data

`Start with random numbers -> look at data -> update random numbers -> look at data -> update`

- `randn`: random numbers from a standard normal distribution with mean of 0 and variance of 1
- `rand` : random numbers from a uniform distribtion

```python
# Create a random tensor of size (3, 4)
# command: torch.rand
random_uniform_tensor = torch.rand(3, 4)
random_uniform_tensor
```

```python
random_normal_tensor = torch.randn(size=(5, 5))
random_normal_tensor
```

```python
# Create a random tensor with similar shape to an image tensor
random_iamge_size_tensor = torch.rand(size=(3, 224, 224)) # color channels, height: cc (R, G, B)
random_iamge_size_tensor.shape, random_iamge_size_tensor.ndim
```

### Creating a range of tensors and tensors-like

returns a 1-D tensor containing a sequence of evenly spaced values within a specified interval (start (inclusive) and end (which is exclusive of the range))

For the range, a step size can be specified to dictate the gap between each value

```python
# Use torch.arange()
one_to_ten = torch.arange(1, 11)
one_to_ten
```

```python
# Use torch.arange()
zero_to_1000 = torch.arange(start=0, end=1000, step=77)
zero_to_1000
```

**_like** suffix is used to create a tensor that mimics the property of another tensor
- `zeros_like`: creates a tensor filled with zeros
- `ones_like`: creates a tensor filled with ones
- `full_like`: creates a tensor filled with a specific scalar value
- `rand_like`: creates a tensor filled with random numbers from a uniform distribution of floating points
- `randn_like`: creates a tensor filled with random numbers from a normal distribution of floating points

```python
ten_zeroes = torch.zeros_like(input=one_to_ten)
ten_zeroes
```

```python
ten_ones = torch.ones_like(one_to_ten)
ten_ones
```

```python
ten_fulls = torch.full_like(one_to_ten, 7)  # specify any number other than one/zero
ten_fulls
```

```python
ten_normals = torch.randn_like(one_to_ten)
ten_normals
```

```python
ten_uniforms = torch.rand_like(one_to_ten)
ten_uniforms
```

The above implementations didn't work because one_to_ten has data type Long/int32 but rand_like and randn_like create of data type float

### Matrix Multiplication

Two main ways of performing multiplication in neural networks and deep learning
1. Scalar Multiplication
   - Multiplying all the elements of a matrix or vector by a scalar
3. Element-wise multiplicaton : a*b
   - Multiplying the correspinding elements of two matrices or vectors of the same shape (also called the **Hadamard product** or **Schur product**)
4. Multiplication (dot product): `a.b` :
   - Multiplying two vectors of the same length
   - To multiply two matrices, the number of columns in the first matrix must be equal to the number of row in the second matrix (To achieve the equal length paradigm) - inner dimensions must match
     - Multiply rows by columns first to fill first row in new matrix
    
#### 1. Inner dimensions rule
* `(3, 2) @ (3, 2)` won't work
* `(2, 3) @ (3, 2)` will work
* `(3, 2) @ (2, 3)` will work

#### 2. Resulting matrix has dimensions of outer dimensions
1. `(3, 2) @ (2, 3)` -> `(3, 3)`
2. `(2, 3) @ (3, 3)` -> `(2, 3)`

http://matrixmultiplication.xyz/

```python
# Element-wise multiplication
print(tensor, "*", tensor)
print(f"Equals: {tensor * tensor}")
```

```python
# Matrix Multiplication
torch.matmul(tensor, tensor)
```

```python
# perform a 3x3 elementwise and matrix multiplication
# create 2 3x3 matrices
first_3by3 = torch.tensor([[2, 3, 4],
                           [6, 5, 4],
                           [9, 8, 1]])

second_3by3 = torch.tensor([[6, 6, 4],
                           [6, 5, 8],
                           [4, 8, 2]])
```

```python
# elementwise
print(first_3by3 * second_3by3)
print()
print(second_3by3 * first_3by3)
```

```python
# matrix multiplication
print(torch.matmul(first_3by3, second_3by3))
print()
print(torch.matmul(second_3by3, first_3by3))
```

**Note**: positioning is important in matrix mulitpication

```python
# @ also works
first_3by3 @ second_3by3
```
